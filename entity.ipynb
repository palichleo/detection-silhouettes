```python
import sys
import numpy as np
import cv2
import mss
from PyQt5 import QtWidgets, QtGui, QtCore
from ultralytics import YOLO

model = YOLO("yolo11x.pt")

class Overlay(QtWidgets.QWidget):
    def __init__(self):
        super().__init__()

        # Supprime les bordures et garde la transparence
        self.setWindowFlags(QtCore.Qt.FramelessWindowHint | QtCore.Qt.WindowStaysOnTopHint | QtCore.Qt.Tool)
        self.setAttribute(QtCore.Qt.WA_TranslucentBackground)
        self.setAttribute(QtCore.Qt.WA_NoSystemBackground, True)
        self.setAttribute(QtCore.Qt.WA_TransparentForMouseEvents)

        # Plein écran
        self.screen = QtWidgets.QApplication.primaryScreen().geometry()
        self.setGeometry(self.screen)

        # Timer pour mise à jour
        self.timer = QtCore.QTimer()
        self.timer.timeout.connect(self.update_frame)
        self.timer.start(30)

        self.boxes = []

    def update_frame(self):
        with mss.mss() as sct:
            monitor = sct.monitors[1]  # écran principal
            screenshot = np.array(sct.grab(monitor))
            frame = cv2.cvtColor(screenshot, cv2.COLOR_BGRA2BGR)

            # YOLO (classe 0 = personne)
            results = model(frame, classes=[0], conf=0.5)
            self.boxes = []

            for box in results[0].boxes.data.tolist():
                x1, y1, x2, y2, conf, cls = box
                self.boxes.append((int(x1), int(y1), int(x2), int(y2)))

            self.update()

    def paintEvent(self, event):
        painter = QtGui.QPainter(self)
        pen = QtGui.QPen(QtGui.QColor(0, 255, 0, 180))
        pen.setWidth(3)
        painter.setPen(pen)

        for (x1, y1, x2, y2) in self.boxes:
            painter.drawRect(x1, y1, x2 - x1, y2 - y1)

if __name__ == "__main__":
    app = QtWidgets.QApplication(sys.argv)
    overlay = Overlay()
    overlay.show()
    sys.exit(app.exec_())
```
