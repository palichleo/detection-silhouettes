```python
import cv2
import numpy as np
import mss
import time

output = "screen_recor.mp4"
fps = 20
duration = 300 # durée en secondes

with mss.mss() as sct:
    monitor = sct.monitors[1]  # Écran principal
    width = monitor["width"]
    height = monitor["height"]
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    out = cv2.VideoWriter(output, fourcc, fps, (width, height))

    print("Enregistrement en cours...")
    start_time = time.time()

    while True:
        img = np.array(sct.grab(monitor))
        frame = cv2.cvtColor(img, cv2.COLOR_BGRA2BGR)
        out.write(frame)

        if time.time() - start_time > duration:
            break

    out.release()
    print(f"Enregistrement terminé → {output}")
```

```python
import cv2
import os

video_path = "screen_record.mp4"
output_dir = "frames"
os.makedirs(output_dir, exist_ok=True)

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
interval = int(fps * 2)  # 1 image toutes les 2 secondes

frame_count = 0
saved = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    if frame_count % interval == 0:
        cv2.imwrite(f"{output_dir}/frame_{saved:04}.jpg", frame)
        saved += 1

    frame_count += 1

cap.release()
print(f"{saved} images extraites dans {output_dir}")
```
